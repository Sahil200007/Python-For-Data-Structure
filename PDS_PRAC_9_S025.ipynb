1. Import a CSV dataset using Pandas.

```python
import pandas as pd
```

```python
df = pd.read_csv("movie_ratings.csv")

print("Original Dataset:")
display(df)
```

2. Find missing values in the dataset.

```python
print("Missing Values:")
print(df.isnull().sum())
```

3. Replace missing numerical values with the mean.

```python
numeric_columns = df.select_dtypes(include='number').columns

for column in numeric_columns:
    df[column] = df[column].fillna(df[column].mean())

print("Missing numerical values replaced with mean.")
```

4. Remove duplicate records.

```python
print("Duplicate records before removal:", df.duplicated().sum())

df = df.drop_duplicates()

print("Duplicate records after removal:", df.duplicated().sum())
```

5. Display the cleaned dataset.

```python
print("Cleaned Dataset:")
display(df)
```
